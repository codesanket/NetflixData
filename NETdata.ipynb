```python
import pandas as pd
import matplotlib.pyplot as plt

data=pd.read_csv('netflix_dataset.csv')

print(data.head())

#count the NAN value
d1=data.isnull().sum()
print(d1)

#drop NA values
d2=data.dropna(subset=['title','type','release_year','genre','country','rating','duration_min','views_millions'])

plt.figure(figsize=(6,4))
type_count=d2['type'].value_counts()
plt.bar(type_count.index,type_count.values,color=['red','green'])
plt.title('data for NET VS TV')
plt.tight_layout()
plt.show()
```
